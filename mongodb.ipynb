**Question 1:** What are the key differences between SQL and NoSQL databases?

SQL databases are relational and store data in structured tables with fixed schemas. They enforce ACID properties and use SQL as a standard query language. Examples include MySQL, PostgreSQL, and Oracle.

NoSQL databases are non-relational and store data in flexible formats such as documents, key-value pairs, wide-columns, or graphs. They are schema-less, meaning records can vary in structure. Examples include MongoDB, Cassandra, and Redis.

- **Schema:** SQL requires a predefined schema; NoSQL allows dynamic, flexible schemas.
- **Scalability:** SQL scales vertically (better hardware); NoSQL scales horizontally (more servers).
- **Data model:** SQL uses tables with rows and columns; NoSQL uses documents, key-value, column-family, or graph structures.
- **Transactions:** SQL has full ACID support by default; NoSQL prioritizes availability, though MongoDB supports multi-document ACID transactions.
- **Use case:** SQL fits structured relational data; NoSQL suits large volumes of unstructured or semi-structured data.

**Question 3:** Explain the concept of collections in MongoDB.

A collection in MongoDB is a grouping of documents, analogous to a table in a relational database. Collections exist within a database and each document is a JSON-like BSON object.

- Collections are schema-less — documents in the same collection do not need to share the same fields.
- They are created automatically when the first document is inserted.
- Each document has a unique `_id` field that acts as its primary key.
- Collections can be indexed to improve query performance.

```javascript
db.Orders.insertOne({
    order_id: "ORD-001",
    customer: "Alice",
    region: "West",
    sales: 750.00
})
```

**Question 4:** How does MongoDB ensure high availability using replication?

MongoDB ensures high availability through **replica sets** — a group of MongoDB instances that all maintain the same dataset.

- A replica set has one **primary node** that receives all write operations and logs them in an oplog.
- **Secondary nodes** continuously replicate the oplog and apply the same changes to their own data.
- If the primary fails, an automatic **election** promotes a secondary to primary within seconds.
- Clients can optionally read from secondaries to spread the read load.

A minimum of 3 nodes is recommended (1 primary, 2 secondaries) so a majority always exists to elect a new primary.

**Question 6:** What is the role of indexes in MongoDB, and how do they improve performance?

Without an index, MongoDB scans every document in a collection to find matches (a collection scan), which is slow for large datasets. Indexes store a sorted reference structure that lets MongoDB jump directly to relevant documents.

Types of indexes:
- **Single field** - index on one field
- **Compound** - index on multiple fields
- **Text** - for full-text search
- **Geospatial** - for location-based queries
- **TTL** - automatically expires documents after a set time

```javascript
db.Orders.createIndex({ region: 1 })
```

The trade-off is that indexes use extra storage and slightly slow down writes, since each index must be updated on every insert, update, or delete.

**Question 7:** Describe the stages of the MongoDB aggregation pipeline.

The aggregation pipeline processes documents through a series of stages, each transforming the data before passing it forward.

- **$match** - Filters documents (like WHERE). Place it early to reduce data volume.
- **$group** - Groups documents by a field and computes aggregates like sum, avg, count, or max.
- **$project** - Reshapes documents by including, excluding, or renaming fields.
- **$sort** - Sorts documents in ascending or descending order.
- **$limit** - Restricts the number of documents passed to the next stage.
- **$skip** - Skips a number of documents, useful for pagination.
- **$lookup** - Performs a left outer join with another collection.
- **$unwind** - Deconstructs an array field so each element becomes a separate document.
- **$addFields** - Adds new computed fields to documents.
- **$count** - Returns the count of documents at that stage.

```javascript
db.Orders.aggregate([
    { $match: { region: "West" } },
    { $group: { _id: "$category", totalSales: { $sum: "$sales" } } },
    { $sort: { totalSales: -1 } }
])
```

**Question 8:** What is sharding in MongoDB? How does it differ from replication?

**Sharding** distributes data across multiple servers (shards) using a shard key. Each shard holds only a portion of the total dataset. It is used to handle datasets and throughput beyond what a single machine can manage.

**Replication** copies the same full dataset to multiple nodes for fault tolerance — if one node goes down, another has the complete data.

The key difference:
- Replication = **same data on multiple nodes** (availability and redundancy)
- Sharding = **different data on different nodes** (horizontal scalability and storage capacity)

In production, MongoDB typically combines both — each shard is itself a replica set.

**Question 10:** What are the ACID properties in the context of MongoDB transactions?

ACID stands for Atomicity, Consistency, Isolation, and Durability. Starting with MongoDB 4.0, multi-document transactions support all four.

- **Atomicity:** All operations in a transaction either complete fully or none of them do. If any step fails, everything is rolled back.
- **Consistency:** A transaction moves the database from one valid state to another, respecting all integrity rules.
- **Isolation:** Changes within a transaction are not visible to other operations until it commits.
- **Durability:** Once committed, changes are permanent even if the system crashes. MongoDB achieves this through journaling and write concern settings.

Before v4.0, ACID was only guaranteed at the single-document level. Multi-document transactions extended this across multiple collections and documents.

**Question 11:** What is the purpose of MongoDB's explain() function?

The `explain()` function shows how MongoDB executes a query — the execution plan. It is used for performance tuning and debugging slow queries.

It reveals:
- Whether an index was used or a full collection scan occurred
- The number of documents examined vs. returned
- The stages the query went through
- Execution time in milliseconds

```javascript
db.Orders.find({ region: "West" }).explain("executionStats")
```

If `totalDocsExamined` is much larger than the number of results, it signals a missing or ineffective index.

**Question 12:** How does MongoDB handle schema validation?

By default MongoDB is schema-less, but it supports optional schema validation using JSON Schema rules defined at the collection level. Rules are applied on insert and update operations.

Validation can enforce required fields, data types, value ranges, and allowed values.

```javascript
db.createCollection("Orders", {
    validator: {
        $jsonSchema: {
            bsonType: "object",
            required: ["order_id", "region", "sales"],
            properties: {
                sales: { bsonType: "double", minimum: 0 },
                region: { bsonType: "string" }
            }
        }
    },
    validationAction: "error"
})
```

Setting `validationAction` to `"error"` rejects invalid documents outright; `"warn"` logs a warning but still allows the insert.

**Question 13:** What is the difference between a primary and a secondary node in a replica set?

**Primary node:**
- The only node that accepts write operations.
- Records all changes in the oplog.
- Handles reads by default unless clients are configured otherwise.

**Secondary node:**
- Continuously replicates data from the primary's oplog.
- Does not accept direct writes.
- Can serve reads if the client sets a read preference like `secondaryPreferred`.
- Participates in elections when the primary becomes unavailable.

If the primary goes down, the secondaries hold an election and one is automatically promoted to primary.

**Question 15:** Explain the concept of embedded documents and when they should be used.

Embedded documents are documents nested inside another document rather than stored in a separate collection.

```javascript
{
    order_id: "ORD-001",
    customer: {
        name: "Alice",
        email: "alice@example.com",
        address: { city: "Seattle", state: "WA" }
    },
    sales: 750
}
```

Use embedding when:
- The nested data is always accessed together with the parent.
- The relationship is one-to-one or one-to-few.
- The embedded data does not need to be queried independently at scale.
- You want to avoid expensive joins.

Avoid embedding when the sub-document grows unboundedly or the same data needs to be shared across many parent documents — in those cases, use a reference to a separate collection instead.

**Question 16:** What is the purpose of MongoDB's $lookup stage in aggregation?

The `$lookup` stage performs a **left outer join** between two collections in the same database. It is used when related data lives in separate collections and needs to be combined.

```javascript
db.Orders.aggregate([
    {
        $lookup: {
            from: "Customers",
            localField: "customer_id",
            foreignField: "_id",
            as: "customerDetails"
        }
    }
])
```

This adds a `customerDetails` array to each order document containing the matching document(s) from the Customers collection. If no match is found, the array is empty.

**Question 18:** What are the advantages of using MongoDB for horizontal scaling?

Horizontal scaling means adding more servers to distribute the workload instead of upgrading a single machine. MongoDB supports this natively through sharding.

- **No single capacity ceiling:** Data and traffic are split across shards, so the system grows by adding more shards.
- **Cost-effective:** Commodity hardware can be used rather than expensive high-end servers.
- **Geographic distribution:** Shards can be placed in different regions for data locality.
- **Transparent to the application:** The mongos router handles query routing, so application code does not need to change.
- **Automatic balancing:** MongoDB redistributes chunks of data evenly across shards as the dataset grows.

This contrasts with vertical scaling, which hits hardware limits and becomes increasingly expensive.

**Question 19:** How do MongoDB transactions differ from SQL transactions?

SQL transactions have supported multi-row, multi-table ACID transactions since the earliest relational databases, and they are a core part of the SQL model.

MongoDB transactions work differently:
- Before v4.0, ACID guarantees only applied at the single-document level. Because MongoDB supports embedding, many scenarios that require multi-table transactions in SQL can be handled with a single document write in MongoDB.
- From v4.0, multi-document ACID transactions were added for replica sets, and from v4.2 for sharded clusters.
- MongoDB transactions are session-based and use `startTransaction`, `commitTransaction`, and `abortTransaction`.
- They are best used sparingly since they carry more overhead than single-document operations.

The main practical difference is that MongoDB's document model often eliminates the need for transactions in the first place.

**Question 20:** What are the main differences between capped collections and regular collections?

**Capped collections** are fixed-size collections that automatically overwrite the oldest documents when the size limit is reached.

Key differences:
- **Fixed size:** Created with a maximum size in bytes and optionally a max document count.
- **Automatic expiry:** Once full, the oldest documents are overwritten automatically.
- **No manual deletion:** Documents cannot be deleted individually; updates that increase document size are not allowed.
- **Insertion order preserved:** Documents are always returned in natural insertion order.

```javascript
db.createCollection("logs", { capped: true, size: 10485760, max: 5000 })
```

Common use cases: application logs, event streams, and any scenario where only the most recent N records matter.

**Question 21:** What is the purpose of the $match stage in MongoDB's aggregation pipeline?

The `$match` stage filters documents in the aggregation pipeline, passing only those that meet the specified conditions to the next stage. It works like a WHERE clause in SQL.

```javascript
db.Orders.aggregate([
    { $match: { region: "West", sales: { $gt: 500 } } },
    { $group: { _id: "$category", total: { $sum: "$sales" } } }
])
```

- Place `$match` as early as possible in the pipeline to reduce the number of documents processed by later stages.
- If an index exists on the matched field, `$match` will use it, making the filter very fast.
- Multiple conditions can be combined using `$and`, `$or`, or standard query operators.

**Question 22:** How can you secure access to a MongoDB database?

- **Enable authentication:** Start MongoDB with `--auth` so users must provide credentials. Create users with specific roles rather than relying on open access.
- **Use RBAC:** Grant users only the permissions they need — for example, read-only access for reporting users.
- **Use TLS/SSL:** Encrypt all connections between clients and the MongoDB server.
- **Restrict network access:** Bind MongoDB to specific IPs using `bindIp`, and use firewalls or VPC rules to block unauthorized access.
- **Keep MongoDB updated:** Apply security patches promptly.
- **Audit logging:** Enable auditing (Enterprise) to track who did what and when.
- **Field-level encryption:** Use client-side field level encryption for sensitive data like PII or payment details.
- Never expose MongoDB directly to the internet without authentication and network controls.

**Question 23:** What is MongoDB's WiredTiger storage engine, and why is it important?

WiredTiger has been MongoDB's default storage engine since version 3.2. It replaced the older MMAPv1 engine and brought significant improvements.

- **Document-level concurrency:** Multiple writes on different documents in the same collection proceed simultaneously, unlike MMAPv1 which used collection-level locking.
- **Compression:** Compresses data on disk using Snappy (default) or zlib, reducing storage by up to 80%.
- **Encryption at rest:** Supports encrypted storage files (Enterprise edition).
- **Checkpointing:** Creates consistent snapshots at regular intervals for crash recovery.
- **Write-ahead logging:** Changes are written to a journal before being applied, ensuring durability.

WiredTiger's concurrency model is the reason MongoDB handles high-throughput workloads well — many reads and writes can happen in parallel without blocking each other.

**Question 9:** Use aggregation to group orders by Region and calculate total sales per region.

````python
from pymongo import MongoClient

client = MongoClient('mongodb://localhost:27017/')
collection = client['superstore']['Orders']

pipeline = [
    {'$group': {'_id': '$Region', 'TotalSales': {'$sum': '$Sales'}}},
    {'$sort': {'TotalSales': -1}}
]

for result in collection.aggregate(pipeline):
    print(result)
````
